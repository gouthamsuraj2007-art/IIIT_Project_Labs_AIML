```python
# ==========================================
# ONLINE FOOD ORDERING SYSTEM
# ==========================================

print("==========================================")
print("       ONLINE FOOD ORDERING SYSTEM")
print("==========================================")

# Customer Details
print("\n----- CUSTOMER DETAILS -----")

name = input("Enter your name: ")
phone = input("Enter your phone number: ")
address = input("Enter your delivery address: ")

# Restaurant Selection
print("\n----- SELECT RESTAURANT -----")
print("1. Spice Hub")
print("2. Food Corner")
print("3. Tasty Treat")

restaurant_choice = int(input("Enter your choice: "))

if restaurant_choice == 1:
    restaurant = "Spice Hub"

elif restaurant_choice == 2:
    restaurant = "Food Corner"

elif restaurant_choice == 3:
    restaurant = "Tasty Treat"

else:
    restaurant = "Spice Hub"
    print("Invalid choice. Spice Hub selected.")

print("\nRestaurant:", restaurant)

# Food Menu
print("\n----- FOOD MENU -----")
print("1. Pizza       - Rs. 200")
print("2. Burger      - Rs. 120")
print("3. Biryani     - Rs. 180")
print("4. Pasta       - Rs. 150")
print("5. Fried Rice  - Rs. 140")
print("6. Sandwich    - Rs. 100")
print("7. French Fries- Rs. 80")
print("8. Coke        - Rs. 50")

total = 0
order = []

# Taking Order
print("\n----- PLACE YOUR ORDER -----")
print("Enter 0 when you have finished ordering.")

while True:

    choice = int(input("\nEnter food number: "))

    if choice == 0:
        break

    if choice >= 1 and choice <= 8:

        quantity = int(input("Enter quantity: "))

        if choice == 1:
            item = "Pizza"
            price = 200

        elif choice == 2:
            item = "Burger"
            price = 120

        elif choice == 3:
            item = "Biryani"
            price = 180

        elif choice == 4:
            item = "Pasta"
            price = 150

        elif choice == 5:
            item = "Fried Rice"
            price = 140

        elif choice == 6:
            item = "Sandwich"
            price = 100

        elif choice == 7:
            item = "French Fries"
            price = 80

        elif choice == 8:
            item = "Coke"
            price = 50

        amount = price * quantity

        total = total + amount

        order.append([item, quantity, price, amount])

        print(item, "added to your order.")

    else:
        print("Invalid food choice.")

# Check if order is empty
if total == 0:

    print("\nYou did not order anything.")
    print("Thank you for visiting!")

else:

    # Order Type
    print("\n----- ORDER TYPE -----")
    print("1. Home Delivery")
    print("2. Takeaway")

    order_type = int(input("Enter your choice: "))

    if order_type == 1:
        order_name = "Home Delivery"
        delivery_charge = 40

    else:
        order_name = "Takeaway"
        delivery_charge = 0

    # Discount
    if total >= 1000:
        discount = total * 20 / 100

    elif total >= 500:
        discount = total * 10 / 100

    else:
        discount = 0

    # GST
    gst = (total - discount) * 5 / 100

    # Final Amount
    final_amount = total - discount + gst + delivery_charge

    # Payment
    print("\n----- PAYMENT METHOD -----")
    print("1. Cash on Delivery")
    print("2. UPI")
    print("3. Debit/Credit Card")

    payment_choice = int(input("Enter payment method: "))

    if payment_choice == 1:
        payment = "Cash on Delivery"

    elif payment_choice == 2:
        payment = "UPI"

    elif payment_choice == 3:
        payment = "Debit/Credit Card"

    else:
        payment = "Cash on Delivery"

    # Order Confirmation
    print("\n==========================================")
    print("          ORDER CONFIRMATION")
    print("==========================================")

    print("Customer Name :", name)
    print("Phone Number  :", phone)
    print("Restaurant    :", restaurant)
    print("Order Type    :", order_name)
    print("Address       :", address)

    print("\n----- ORDER DETAILS -----")

    print("Item\t\tQty\tPrice\tAmount")

    for food in order:
        print(food[0], "\t\t", food[1],
              "\tRs.", food[2],
              "\tRs.", food[3])

    print("------------------------------------------")

    print("Food Total     : Rs.", total)
    print("Discount       : Rs.", discount)
    print("GST (5%)       : Rs.", gst)
    print("Delivery Charge: Rs.", delivery_charge)

    print("------------------------------------------")

    print("FINAL AMOUNT   : Rs.", final_amount)
    print("Payment Method :", payment)

    print("==========================================")
    print("       ORDER PLACED SUCCESSFULLY!")
    print("==========================================")

    # Rating
    print("\n----- RATE YOUR ORDER -----")
    rating = int(input("Give rating from 1 to 5: "))

    if rating == 5:
        print("Excellent! Thank you for your feedback.")

    elif rating >= 3:
        print("Thank you for your feedback.")

    else:
        print("We will try to improve our service.")

    print("\nThank you for ordering from", restaurant)
    print("Have a great day!")
```
